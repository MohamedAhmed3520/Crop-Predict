## Import Important Libraries

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import  StandardScaler , LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score , confusion_matrix , classification_report , roc_auc_score , f1_score , precision_score , recall_score , mean_squared_error , r2_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression , LinearRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from imblearn.over_sampling import SMOTE
import joblib
from sklearn.model_selection import GridSearchCV , RandomizedSearchCV
```

## Encode the Label Column to Number to be easy to Predict

```python
crop['label'] = LabelEncoder().fit_transform(crop['label'])
```

```python
crop.head(5)
```

```python
crop['label'].value_counts()
```

## Let's Start with Splitting data in x , y

```python
x = crop.drop('label',axis=1)
y = crop['label']
```

```python
y.value_counts(normalize=True)
```

## Splitted it into train and test

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2)
```

```python
x_tr , x_val , y_tr , y_val = train_test_split(x_train,y_train,test_size=0.2)
```

```python
std = StandardScaler()

x_tr = std.fit_transform(x_tr)
x_val = std.transform(x_val)
x_test = std.transform(x_test)
```

## Let's Try all Models and we will Decide which is The Best

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier()
dt.fit(x_tr,y_tr)
dt.score(x_val,y_val)
```

```python
print(classification_report(y_val,dt.predict(x_val)))
```

```python
rfc = RandomForestClassifier()
rfc.fit(x_tr,y_tr)
rfc.score(x_val,y_val)
```

```python
print(classification_report(y_val,rfc.predict(x_val)))
```

```python
lin_svc = SVC(kernel='linear')
lin_svc.fit(x_tr,y_tr)
lin_svc.score(x_val,y_val)
```

```python
print(classification_report(y_val,lin_svc.predict(x_val)))
```

```python
poly_svc = SVC(kernel='poly')
poly_svc.fit(x_tr,y_tr)
poly_svc.score(x_val,y_val)
```

```python
print(classification_report(y_val,poly_svc.predict(x_val)))
```

```python
poly_svc = SVC(kernel='rbf')
poly_svc.fit(x_tr,y_tr)
poly_svc.score(x_val,y_val)
```

```python
print(classification_report(y_val,poly_svc.predict(x_val)))
```

```python
naive = GaussianNB()
naive.fit(x_tr,y_tr)
naive.score(x_val,y_val)
```

```python
print(classification_report(y_val,naive.predict(x_val)))
```

```python
KNN = KNeighborsClassifier(n_neighbors=20)
KNN.fit(x_tr,y_tr)
KNN.score(x_val,y_val)
```

```python
print(classification_report(y_val,KNN.predict(x_val)))
```

```python
Log = LogisticRegression()
Log.fit(x_tr,y_tr)
Log.score(x_val,y_val)
```

```python
print(mean_squared_error(y_val,Log.predict(x_val)))
print(r2_score(y_val,Log.predict(x_val)))
```

```python
linear = LinearRegression()
linear.fit(x_tr,y_tr)
linear.score(x_val,y_val)
```

```python
print(mean_squared_error(y_val,linear.predict(x_val)))
print(r2_score(y_val,linear.predict(x_val)))
```

```python
RFC_Grid.fit(x_tr,y_tr)
```
